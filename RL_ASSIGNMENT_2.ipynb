```python
# Epsilon-Greedy Strategy for Multi-Armed Bandit

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 3

# Exploration probability
epsilon = 0.2

# Part B - Initialize N, R and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# Fixed r values to match the calculation
r_values = [0.8, 0.1, 0.05]

# Fixed machines for exploration
explore_machines = ['B', 'A']

# Rewards
rewards = {
    'A': [3, 4],
    'B': [6],
    'C': []
}

reward_index = {'A': 0, 'B': 0, 'C': 0}
explore_index = 0

print("=== Epsilon-Greedy Multi-Armed Bandit ===")

for play in range(N_play):

    # Generate r value
    r = r_values[play]

    # Part C - Epsilon-Greedy Decision
    if r < epsilon:
        # Exploration
        decision = "Exploration"
        selected_machine = explore_machines[explore_index]
        explore_index += 1

    else:
        # Exploitation
        decision = "Exploitation"

        # Select machine with highest Q-value
        max_q = max(Q.values())

        for machine in machines:
            if Q[machine] == max_q:
                selected_machine = machine
                break

    # Get reward
    reward = rewards[selected_machine][reward_index[selected_machine]]
    reward_index[selected_machine] += 1

    print(f"\nPlay {play + 1}")
    print(f"Random number: {r}")
    print(f"Decision: {decision}")
    print(f"Selected Machine: {selected_machine}")
    print(f"Reward: {reward}")

    # Part D - Update N
    N[selected_machine] += 1

    # Update total reward
    R[selected_machine] += reward

    # Incremental Sample-Average Formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    print(f"Updated Q({selected_machine}) = {Q[selected_machine]:.2f}")


# Part E - Display Results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tQ")
print("--------------------------------")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]:.2f}\t"
        f"{Q[machine]:.2f}"
    )

# Analysis
print("\n========== ANALYSIS ==========")

for machine in machines:
    print(
        f"Machine {machine}: "
        f"Played {N[machine]} times, "
        f"Total Reward = {R[machine]:.2f}, "
        f"Estimated Q-value = {Q[machine]:.2f}"
    )

best_machine = max(machines, key=lambda x: Q[x])

print(f"\nBest machine based on Q-value: {best_machine}")
print(f"Highest estimated reward (Q-value): {Q[best_machine]:.2f}")
```
